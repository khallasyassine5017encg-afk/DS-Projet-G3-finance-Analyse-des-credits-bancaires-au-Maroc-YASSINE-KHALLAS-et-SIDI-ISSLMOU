# 4. Analyse des données et interprétation

4.1. Analyse de l'évolution du crédit bancaire

Objectif : étudier l'évolution du montant total des crédits accordés par les banques marocaines entre 2001 et 2026, afin de comprendre la progression du financement bancaire au cours de cette période.



Requête 1 : afficher le montant du crédit bancaire au début et à la fin de la période.

```python
# Récupérer la ligne du crédit bancaire
credit = df.loc[
    df.iloc[:, 0].astype(str).str.strip() == 'Crédit bancaire'
].iloc[0, 1:]

# Convertir les montants en valeurs numériques
credit = pd.to_numeric(credit, errors='coerce').dropna()

# Afficher les résultats
print("Premier montant :", round(credit.iloc[0], 2), "MDH")
print("Dernier montant :", round(credit.iloc[-1], 2), "MDH")

print("Augmentation totale :",
      round(credit.iloc[-1] - credit.iloc[0], 2), "MDH")
```

L'analyse de l'évolution du crédit bancaire au Maroc montre une augmentation importante de l'encours total des crédits durant la période étudiée. En décembre 2001, le montant du crédit bancaire était de 204 873,15 millions de dirhams, tandis que la dernière observation disponible atteint 1 333 309,80 millions de dirhams.

Cela représente une augmentation absolue de 1 128 436,65 millions de dirhams. Cette évolution traduit une expansion importante du financement bancaire au Maroc sur le long terme.

Cette progression peut notamment être associée à l'augmentation des besoins de financement des entreprises, au développement du marché immobilier, à la croissance des besoins des ménages et à l'évolution de l'activité économique.

Elle met également en évidence le rôle essentiel du système bancaire dans la mobilisation et la distribution des financements destinés à l'économie nationale.

Toutefois, cette augmentation de l'encours ne signifie pas automatiquement une amélioration de la situation économique. Il faut également prendre en considération l'inflation, l'évolution de la production économique et la qualité des crédits accordés.

**4.2. Calculer le taux d'évolution global du crédit bancaire**

Objectif : mesurer en pourcentage l'augmentation du crédit bancaire entre décembre 2001 et la dernière observation disponible.


```python
montant_initial = credit.iloc[0]
montant_final = credit.iloc[-1]

taux_evolution = (
    (montant_final - montant_initial) / montant_initial
) * 100

print("Taux d'évolution global :",
      round(taux_evolution, 2), "%")
```

**Interprétation économique :**

Le taux d'évolution permet de mesurer l'ampleur de la progression du crédit bancaire par rapport à son niveau initial.

Dans notre étude, le crédit bancaire a connu une augmentation très importante depuis décembre 2001. Cette évolution montre que les financements bancaires occupent une place de plus en plus importante dans l'économie marocaine.

Cette progression peut s'expliquer par le développement du système bancaire, l'élargissement de l'accès au financement et l'augmentation des besoins des différents agents économiques.

Elle souligne également l'importance du crédit bancaire comme instrument de financement de l'investissement, du logement et de la consommation.

**4.3. Identifier les catégories de crédits les plus importantes**

Objectif : déterminer quelles catégories représentent les montants les plus élevés dans le financement bancaire marocain.

Requête 3 : classer les catégories de crédits selon leur montant à la dernière date disponible.

```python
# Récupérer la dernière colonne de données
derniers_montants = df.iloc[:, 1:].apply(
    pd.to_numeric, errors='coerce'
).iloc[:, -1]

# Associer les catégories à leurs montants
classement = pd.DataFrame({
    'Categorie': df.iloc[:, 0],
    'Montant_MDH': derniers_montants
})

# Supprimer les valeurs manquantes et le crédit bancaire total
classement = classement.dropna()
classement = classement[
    classement['Categorie'] != 'Crédit bancaire'
]

# Trier les catégories par ordre décroissant
classement = classement.sort_values(
    by='Montant_MDH', ascending=False
)

display(classement)
```

Crédits à caractère financier	172 413,53

Créances en souffrance	105 469,95

Crédits aux promoteurs immobiliers	65 446,60

Crédits à la consommation	64 096,37

Le classement montre que les crédits à l'équipement occupent la première position, avec un montant de 356 664,66 MDH. Ces crédits permettent notamment aux entreprises de financer l'acquisition de machines, de matériels et d'autres moyens nécessaires au développement de leurs activités et à l'amélioration de leur capacité de production.

En deuxième position, nous trouvons les crédits immobiliers, avec 332 603,51 MDH. Leur importance montre que le secteur immobilier représente également un domaine majeur du financement bancaire. Ces financements répondent notamment aux besoins liés au logement et aux activités immobilières.

Les crédits de trésorerie occupent la troisième position, avec 275 797,90 MDH. Ils jouent un rôle essentiel dans le fonctionnement des entreprises, puisqu'ils leur permettent de couvrir leurs besoins financiers à court terme, comme le paiement des fournisseurs, des salaires et des dépenses courantes.

Par ailleurs, les crédits à l'habitat représentent 260 520,15 MDH. Ce montant confirme l'importance du financement du logement dans le système bancaire marocain. Il faut toutefois préciser que cette catégorie est incluse dans les crédits immobiliers et ne doit donc pas être additionnée une deuxième fois au total.

En revanche, les crédits à la consommation représentent 64 096,37 MDH, un montant relativement faible par rapport aux crédits à l'équipement et aux crédits immobiliers. Cela montre que, dans cette répartition, les financements destinés à l'investissement et à l'immobilier occupent une place plus importante que ceux consacrés directement à la consommation.

**4.9. Analyse de la part des crédits à l'habitat dans les crédits immobiliers**

Objectif : déterminer quelle proportion des crédits immobiliers est consacrée au financement de l'habitat.

```python
immobilier = df.loc[
    df.iloc[:, 0].astype(str).str.strip()
    == "Crédits immobiliers"
].iloc[0, 1:]

habitat = df.loc[
    df.iloc[:, 0].astype(str).str.strip()
    == "Crédits à l'habitat"
].iloc[0, 1:]

immobilier = pd.to_numeric(immobilier, errors='coerce').dropna()
habitat = pd.to_numeric(habitat, errors='coerce').dropna()

part_habitat = (
    habitat.iloc[-1] / immobilier.iloc[-1]
) * 100

print("Crédits immobiliers :", round(immobilier.iloc[-1], 2), "MDH")
print("Crédits à l'habitat :", round(habitat.iloc[-1], 2), "MDH")
print("Part des crédits à l'habitat :", round(part_habitat, 2), "%")
```

Les résultats montrent que les crédits à l'habitat représentent 78,33 % de l'ensemble des crédits immobiliers, ce qui signifie que la majeure partie du financement immobilier bancaire est destinée au logement. Cette proportion élevée souligne l'importance du financement de l'habitat au Maroc et le rôle des banques dans l'accompagnement des ménages pour l'acquisition ou la construction de logements. Les 21,67 % restants correspondent aux autres composantes du financement immobilier, notamment les crédits aux promoteurs immobiliers. Ainsi, le financement du logement constitue la principale composante des crédits immobiliers au Maroc.
